## Converting Python List to Numpy array
- Syntax: np.array(a)
- dtype - To find Every Element DataType
- Converting python List to numpy array with specific Datatype
- a=np.array(a,dtype=np.int8)
- np.array(a,dtype=str_)

NOTE:
- above example [210 ,130] = ithu out of range (allowed range(-128 to 127)
1. 210 (210-256) = -46 ### Value > 127 => subtract 256
2. -130 (-130+256) = 126 ### Value < -128 => Add 256

## Typecasting 
- a.astype(np.int32)


```python
import numpy as np
```

```python
a=[1,2,3,4]
print(a)
print(type(a))
a=np.array(a)
print(a)
print(type(a))
print(a.dtype)
```

```python
a=['Cat',1,1.2,True]
print(a)
print(type(a))
a=np.array(a)
print(a)
print(type(a))
print(a.dtype)
```

```python
# Creating array with specific datatype
a=[1,2,3,4]
print(a)
a=np.array(a)
print(a.dtype)
np.array(a,dtype=np.int8)
```
