```python
import numpy as np
x=np.array([1,2,3,4])
y=np.array([2,4,6,8])
plt.plot(x,y,color="pink")
```

```python
plt.plot(x,y,color="green",marker='^',linestyle='dotted',linewidth=3,markerfacecolor='yellow',markeredgecolor="blue",markersize=20)
```

```python
sample_property=dict(marker='^',linestyle='dotted',linewidth=3,markerfacecolor='purple',markeredgecolor="green",markersize=20)
plt.plot(x,y,**sample_property)
```

```python
plt.plot(x,y,**sample_property)
plt.xlabel("x-axis",color="red")
plt.ylabel("y-axis",color="blue")
plt.grid(color="blue")
plt.show()
```
