# Boxplot 
- To compare between different section in same group
- eg: A<B<C<D section in standard 6th


```python
import numpy as np
```

np.random.seed(1)
np.random.normal(175,7,100) # 5,7,9,10
np.random.normal(175,7,100) # 5,7,9,10

np.random.seed(2)
np.random.normal(175,7,100) # 15,17,19,20
np.random.normal(175,7,100) # 15,17,19,20

```python
np.random.seed(2)
score=np.random.normal(175,7,100)
plt.title("Box plot",color="RED")
plt.boxplot(score,boxprops=dict(color='pink'))
plt.show()
```
