```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset("tips")
print(tips.head())
```

```python
#1. Scatter Plot — Relationship between two values
# Purpose: Find relationship between two numerical columns.
# Use case: Bill amount vs Tip amount.
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.show()
```

```python
# 2. Line Plot — Trend over time or sequence
# Purpose: Show increasing/decreasing trends.
# Use case: Sales trend, Temperature trend.
sns.lineplot(data=tips, x="size", y="total_bill")
plt.show()
```

```python
# 3. Bar Plot — Compare categories
# Purpose: Compare average values of categories.
# Use case: Average sales by day.
# sns.barplot(data=tips, x="day", y="total_bill",hue='day',palette="dark")
sns.barplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
# 4. Count Plot — Count category values
# Purpose: Count occurrences of each category.
# Use case: Number of orders each day.
sns.countplot(data=tips, x="day")
plt.show()
```

```python
# 5. Histogram — Data Distribution
# Purpose: Show frequency distribution.
# Use case: Student marks distribution.
sns.histplot(data=tips, x="total_bill", bins=10)
plt.show()
```

```python
# 6. Box Plot — Find Outliers
# Purpose: Identify minimum, maximum, median, and outliers.
# Use case: Salary analysis, Exam marks.
sns.boxplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
# 7. Violin Plot — Distribution + Density
# Purpose: Show distribution shape and spread.
# Use case: Compare distribution across categories.
sns.set_palette('dark')
sns.violinplot(data=tips, x="day", y="total_bill")

plt.show()
```

```python
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    palette="dark",
    hue="day",
    legend=True
)

plt.show()
```

```python
# 8. Pair Plot — Relationship between all numeric columns
# Purpose: Compare every numeric column together.
# Use case: Machine Learning EDA.
iris = sns.load_dataset("iris")
sns.pairplot(iris)
sns.set_theme(style="white")
plt.show()
```

```python
# 9. Heatmap — Correlation Matrix
# Purpose: Show correlation between columns.
# Use case: Feature relationship in data analytics.
corr = tips.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
```
