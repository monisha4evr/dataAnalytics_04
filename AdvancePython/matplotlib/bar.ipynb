```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
x=np.array([2020,2021,2022,2023])
y=np.array([10000,2000,15000,20000])
plt.title("Revenue",color="Red")
plt.bar(x,y,color="purple")
```

```python
x=np.array([2020,2021,2022,2023])
y=np.array([10000,2000,15000,20000])
plt.barh(x,y,color="purple")
```

```python
x=['apple','banana','carrot','drygrapes']
y=np.array([10000,2000,15000,20000])
plt.pie(y,labels=x,autopct="%1f%%",colors=["red","yellow","orange","purple"],shadow=True,explode=[0,0,0.5,0],)
plt.show()
```
