```python
import numpy as np
```

```python
a=[1,2,3,4]
print(a+2)
```

```python
a=np.array([1,2,3,4])
print(a+2)
```

```python
a = np.array([[1,2,3],
              [4,5,6]])
print(a)
print("Shape",a.shape)
print("Dimension",a.ndim)
print("Addition: \n",a+3)
print("Sub: \n",a-3)
print("Multiplication: \n",a*3)
print("Divide\n",a/3)
print("Floor Dividion\n",a//3)
print("Exponentials\n",a**2)
```

# Matrix Multiplication

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print(A@B)
print("\n",np.matmul(A,B))
```

```python
print(np.prod(A))
```
