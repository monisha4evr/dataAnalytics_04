# Typecasting
**convert one datatype to another**

```python
a=[15,25,35,45,55]
a=np.array(a)
print(a)
print(a.dtype)
a=a.astype(dtype=np.int8)
print(a.dtype)
```

```python
a=[15,25,35,45,55]
a=np.array(a)
print(a)
print(a.dtype)
a=a.astype(dtype=np.float16)
print(a)
print(a.dtype)
```

```python
a=[15,25,135,45,155]
a=np.array(a)
print(a)
print(a.dtype)
a=a.astype(dtype=np.int8) # typeconversion
print(a)
print(a.dtype)

# NOTE: 155 - 256 = -101
# 135 - 256 = -121
```

```python
print(256 - 155)
```

```python
a=[-215,25,-205,45,15]
a=np.array(a)
print(a)
print(a.dtype)
a=a.astype(dtype=np.int8) # typeconversion
print(a)
print(a.dtype)
# NOTE: -215+256 = 41
# NOTE: -205+256 = 45
```
