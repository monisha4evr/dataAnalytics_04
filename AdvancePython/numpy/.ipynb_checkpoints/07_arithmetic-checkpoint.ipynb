```python
import numpy as np
```

```python
a = np.array([[1,2,3],
              [4,5,6]])
print(a + 2)
print(a - 2)
print(a * 2)
print(a / 2)
print(a % 2)
print(a ** 2)
```

# Matrix Multiplication

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

result = A @ B

print(result)
```

```python
np.matmul(A, B)
```
