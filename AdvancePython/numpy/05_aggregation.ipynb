```python
import numpy as np
```

```python
#Aggregation Functions (sum,mean,median,max,min,std,var,argmax)
```

```python
arr=[1,4,7,9,3]
print(sum(arr))
print(arr)
```

```python
arr=np.array([1,4,7,9,3])
print(arr)
print(np.sum(arr))
```

```python
arr=np.array([1,4,7,9,3])
print(arr)
print(np.min(arr))
```

```python
arr=np.array([1,4,7,9,3])
print(arr)
print(np.max(arr))
```

```python
arr=np.array([1,4,7,9,3])
print(arr)
print(np.argmax(arr))
```

```python
arr=np.array([1,4,7,9,3])
print(arr)
print(np.argmin(arr))
```

```python
#mean,median,std,var
arr=np.array([1,4,7,9,3,5])
print(np.mean(arr))
print(np.median(arr)) # 1,3,4,5,7,9 => 9/2
print(np.std(arr))
print(np.var(arr))
```
