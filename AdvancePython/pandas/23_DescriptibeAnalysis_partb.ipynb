# Relationship Analysis

## 1. Covariance - how one Variable move with Other Variable in terms of Direction
- height incr -> weight incr => Positive
- height incr -> weight decr => Negative

```python
df['Horsepower'].cov(df['weight'])
# just check positive or negative 
# (+) both increase
```

## 2. Correlation - how two variables related together? in Values & direction

```python
df['Horsepower'].corr(df['weight'])*100
# both incr
# to check with percentage ( *100)
```

## 3. Correlation with - compare a variable with a Dataframe

```python
x=df.select_dtypes(include='number').drop(columns='weight')
y=df['weight']
```

```python
x.corrwith(y)
```

# Cumulative Operation

```python
df_courses=pd.DataFrame([['22C01','C#',6000],['22C02','C++',9000],['22C03','Dart',6000],['22C04','Flutter',18000],['22C05','java',6000]])
df_courses.columns=["ID","Name","fee"]
df_courses
```

```python
df_payment=pd.DataFrame([['22C001','2022-01-02','22S001','22C15',1000],['22C002','2022-01-02','22S001','22C12',8000],['22C003','2022-01-09','22S002','22C02',1000],['22C004','2022-01-09','22S002','22C03',7000],['22C005','2022-01-09','22S002','22C02',5000]])
df_payment.columns=['Transaction','Data','Student','Course','Paid']
df_payment
```

## Running Sum

```python
# Find the Total Sales
df_payment['Running Sum']=df_payment['Paid'].cumsum()
df_payment
```

```python
# Running Max | Cumulative MAX
```

```python
df_payment['Max sales']=df_payment['Paid'].cummax()
df_payment
```

```python
# Running Min | Cumulative Min
```

```python
df_payment['Min sales']=df_payment['Paid'].cummin()
df_payment
```
