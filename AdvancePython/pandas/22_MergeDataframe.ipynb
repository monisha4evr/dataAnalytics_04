```python
import pandas as pd
```

```python
df_courses=pd.DataFrame([['22C01','C#',6000],['22C02','C++',9000],['22C03','Dart',6000],['22C04','Flutter',18000],['22C05','java',6000]])
df_courses.columns=["ID","Name","fee"]
df_courses
```

```python
df_payment=pd.DataFrame([['22C001','2022-01-02','22S001','22C15',1000],['22C002','2022-01-02','22S001','22C12',8000],['22C003','2022-01-09','22S002','22C02',1000],['22C004','2022-01-09','22S002','22C03',7000],['22C005','2022-01-09','22S002','22C02',5000]])
```

```python
df_payment.columns=['Transaction','Data','Student','Course','Paid']
```

```python
df_payment
```

## collect all Transaction for each Courses

```python
df_courses.merge(df_payment,left_on='ID',right_on='Course') # default inner
```

## Merge with left join

```python
df_courses.merge(df_payment,how="left",left_on='ID',right_on='Course')
```

```python
df_courses[['ID','Name']].merge(df_payment[['Course','Paid']],left_on='ID',right_on='Course')
```

```python
## Merge with Right join
```

**Replace Course ID with Course Name in Payment Table**

```python
df_courses.merge(df_payment,how='right',left_on='ID',right_on='Course')
```

```python
df_courses[['ID','Name']].merge(df_payment,how='right',left_on='ID',right_on='Course').drop(['ID','Course'],axis=1)
```

## Merge with Inner join

Get only sold Course and related Transaction

```python
df_courses.merge(df_payment,how="inner",left_on='ID',right_on='Course')
```

## Merge with Outer join

```python
df_courses.merge(df_payment,how="outer",left_on='ID',right_on='Course')
```
