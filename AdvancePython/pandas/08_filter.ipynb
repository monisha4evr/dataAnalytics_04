```python
# Sample DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Email': ['alice@gmail.com', 'bob@yahoo.com', 'charlie@gmail.com', 'david@outlook.com']}
df = pd.DataFrame(data)

# Filter rows where Email ends with 'gmail.com'
# The regex pattern matches text ending with 'gmail.com'
gmail_df = df[df['Email'].str.contains(r'gmail\.com$', regex=True)]
print(gmail_df)
```
