```python
import pandas as pd
```

```python
# concatenate (join vertical)
```

```python
df1=pd.DataFrame(
    [
        [1,5],
        [5,2]
    ]
)
df1.columns=['A','B']
df1
```

```python
df2=pd.DataFrame([
    [12,5],
    [1,4]
])
df2.columns=['A','B']
df2
```

```python
##  Concatenate with same Column Names
```

```python
pd.concat([df1,df2])
```

```python
## Concatenate with Various Column Names
```

```python
df3=pd.DataFrame(
    [[12,5],
    [1,4]]
)
df3.columns=['B','C']
df3
```

```python
pd.concat([df1,df3])
```

## Joining with Various Column Names (horizonal)

```python
df1.join(df2,lsuffix="_one",rsuffix="_two")
```

## Joining with same Column Names

```python
df1.join(df3,lsuffix="_one",rsuffix="_two")
```
