# Align with Left join
- Aligning a DataFrame in Pandas means matching rows and Columns of two data Structures based on their labels(row index and column names)
- dataframe 1 (left) and dataframe 2 (right)  
- **Align-left** - Considers labels only available in Left Tables
- abc - cdef => abc c

```python
df1=pd.DataFrame([[1,5,9],[5,2,10],[30,90,2]])
df1.columns=['A','B','C']
df1
```

```python
df2=pd.DataFrame([[9,10,5,2],[1,2,3,4],[9,7,4,1],[3,6,6,4]])
df2.columns=['C','D','E','F']
df2
```

```python
## 1. Align Based on Columns Label
```

```python
left, right=df1.align(df2,join='left')
```

```python
left
```

```python
right
```

```python
## 2.  Align Based on Row Index
```

```python
left, right=df1.align(df2,join='left',axis=0)
```

```python
left
```

```python
right
```

```python
# Align with Right join
```

```python
left , right=df1.align(df2,join='right')
```

```python
left
```

```python
right
```

```python
left , right=df1.align(df2,join='right' , axis=0)
```

```python
left
```

```python
right
```

```python
left , right =df1.align(df2,join='inner')
```
