# Skew
Skewness-ன் 3 முக்கிய வகைகள்:
- Positive Skew (வலது பக்கம் சாய்ந்தது / Right-skewed):
    - விளக்கம்: வரைபடத்தின் நீண்ட வால் (long tail) வலது பக்கமாக நீண்டு இருக்கும்.
    - பெரும்பாலான டேட்டாக்கள் இடது பக்கத்தில் (சிறிய மதிப்புகளில்) குவிந்திருக்கும்.
    - விகிதம்: இதில் சராசரி (Mean) ஆனது மீடியனை (Median) விட அதிகமாக இருக்கும் (Mean > Median).
    - **நிஜ உலக உதாரணம்:** மக்களின் வருமானம் (Income).
   - ஒரு நாட்டின் பெரும்பாலான மக்கள் நடுத்தர அல்லது குறைந்த வருமானமே ஈட்டுவர் (இடது பக்க குவிப்பு). ஆனால், ஒரு சில கோடீஸ்வரர்களின் அதீத வருமானம் இந்த வரைபடத்தை வலது பக்கமாக இழுத்து ஒரு நீண்ட வாலை உருவாக்கும்.
- Negative Skew (இடது பக்கம் சாய்ந்தது / Left-skewed):
   - விளக்கம்: வரைபடத்தின் நீண்ட வால் (long tail) இடது பக்கமாக நீண்டு இருக்கும்.
   - பெரும்பாலான டேட்டாக்கள் வலது பக்கத்தில் (பெரிய மதிப்புகளில்) குவிந்திருக்கும்.
   - விகிதம்: இதில் சராசரி (Mean) ஆனது மீடியனை (Median) விட குறைவாக இருக்கும் (Mean < Median).
   - **நிஜ உலக உதாரணம்:** மனிதர்கள் ஓய்வு பெறும் வயது (Retirement Age).
   - பெரும்பாலான மக்கள் தங்கள் 60 அல்லது 65 வயதில் ஓய்வு பெறுவார்கள் (வலது பக்க குவிப்பு).
   - ஆனால், மிகச் சிலரே தங்களின் 30 அல்லது 40 வயதிலேயே ஓய்வு பெறுவர்.
   - இந்த ஆரம்பகால ஓய்வு வரைபடத்தை இடது பக்கமாக இழுக்கும்.
- Zero Skew (சமச்சீர் நிலை / Symmetric):
   - விளக்கம்: வரைபடம் இடதும் வலதும் சரியாக சமமாக இருக்கும். எந்தப் பக்கமும் சாயாது.
   - விகிதம்: இதில் Mean, Median, Mode ஆகிய மூன்றும் ஒரே புள்ளியில் இருக்கும் (Mean = Median = Mode).
   - **நிஜ உலக உதாரணம்:** மனிதர்களின் உயரம் (Height). ஒரு குறிப்பிட்ட வயதில் உள்ள மனிதர்களின் உயரத்தை அளந்தால், பெரும்பாலானோர் சராசரி உயரத்திலும், மிகச் சிலரே மிகக் குட்டையாகவோ அல்லது மிக உயரமகவோ இருப்பார்கள்.

```python
#skewness how data is distributed
df['weight'].skew()
# near to 0 distributed Evenly 
# (+ ) Right la skew
# (-) left side la skew
```

# kurtosis
- 1. Leptokurtic (உயரமான கோபுரம் & தடிமனான வால் - High Kurtosis):
    - விளக்கம்: இதன் நடுப்பகுதி மிகவும் உயரமாக இருக்கும். அதே சமயம், அதன் இரு பக்கவாட்டு வால்கள் (tails) மிகவும் தடிமனாக இருக்கும்.
    - பொருள்: இதில் எதிர்பாராத, மிக விசித்திரமான சம்பவங்கள் (extreme outliers) அதிகமாக நடக்கும்.
    - உதாரணம்: பங்குச் சந்தை (Stock Market). பொதுவாகப் பங்கு விலை பெரிய மாற்றம் இல்லாமல் இருக்கும் (உயரமான நடுப்பகுதி). 
        - ஆனால், திடீரென ஒரு நாள் பெரும் வீழ்ச்சியோ அல்லது மிகப்பெரிய லாபமோ (Fat tails) ஏற்படும்
- 2.Platykurtic (தட்டையான வடிவம் & மெல்லிய வால் - Low Kurtosis)
  - விளக்கம்: இது பார்க்க மிகவும் தட்டையாக (flat) இருக்கும். இதன் பக்கவாட்டு வால்கள் மிகவும் மெலிதாக இருக்கும் அல்லது வால்களே இருக்காது.
  - பொருள்: இதில் அதீத விளிம்பு நிலை மாற்றங்கள் (outliers) இருக்காது. டேட்டா அனைத்தும் பரவலாக சமமாக இருக்கும்.
  - உதாரணம்: ஒரு பகடையை (Dice) உருட்டுவது. நீங்கள் 1000 முறை பகடையை உருட்டினால் 1, 2, 3, 4, 5, 6 ஆகிய அனைத்து எண்களும் தோராயமாக சமமான முறையே விழும். இதனால் வரைபடம் தட்டையாக இருக்கும். 7 விழ வாய்ப்பே இல்லை என்பதால் வால்கள் இருக்காது
- Mesokurtic (சாதாரண நிலை - Normal Distribution):
  - விளக்கம்: இது சரியான அளவில் இருக்கும் ஒரு பெல் கர்வ் (Normal bell curve).
  - பொருள்: இதன் Kurtosis மதிப்பு சரியாக 3 ஆக இருக்கும் (Excess Kurtosis = 0)

```python
# kurtosis
df['weight'].kurt() 
# (-)  Less
# (0) no 
# (+) High Kurtosis
```
