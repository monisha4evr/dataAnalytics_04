## over Sampling
- total record 500 need 1000 records
- **Sampling without replacement** Dont replace the item from the population wheever you pick it up. So the same item cannot be picked more than once in the sample
- **Sampling with Replacement** Replace the item from the population wheever you pick it up. so htere is a change of picking the same item again in the sample

```python
df.sample(n=1000,replace=True)
```

```python
df.sample(n=1000,replace=True).sort_values(['car','MPO']) .head(60)
```
