# Function
- Custom Function (series,Dataframe) => .apply() eg: df.apply(custom_func)
  - NOTE: for dataframe need to mention pd.Series in function eg(def convert_function(column:pd.Series):)
  - column.str.lower().str.replace(' ','') for Chaining
    - assign() is used to create or update columns,
    - while apply() is used to modify values using a function
- Aggregate Function
    - sum,product,prod,mean,median,min,max,count,size,std,var
      - eg: for Dataframe: df.sum(numeric_only=True)
    - groupby eg: df['weight'].groupby(df['origin']).max()
        - df.groupby('origin').max(numeric_only=True)
        - df.groupby('origin')['weight'].max(numeric_only=True)
- unique()
- Map
  - interate with Every integer Element
  - apply
- Lambda
- Pipe (series / continous operation) .pipe().pipe()

```python
df=pd.read_csv('../../../car_data_500.csv').head()

def convert_to_kw(hp):
    return hp*0.745

def fix_decimals(value):
    return round(value,2)
    
df['pipe_kw_dec']=df['Horsepower'].pipe(convert_to_kw).pipe(fix_decimals)
```
